```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

print("Cooking up 25,000 restaurant transaction records... please wait.")

np.random.seed(42)
num_orders = 25000

# Operational locations
locations = ['Ravet, Pune', 'Koregaon Park, Pune', 'FC Road, Pune', 'Viman Nagar, Pune']

# Menu Items with Cost of Goods Sold (COGS), Sale Price, and Target Kitchen Prep Time (minutes)
menu_items = {
    'Premium Maharaja Thali': {'cost': 250, 'price': 600, 'target_time': 20},
    'Irani Chai & Bun Maska': {'cost': 30, 'price': 120, 'target_time': 5},
    'Mutton Seekh Kebabs': {'cost': 140, 'price': 350, 'target_time': 15},
    'Spicy Miso Ramen': {'cost': 180, 'price': 450, 'target_time': 12},
    'Berry Iced Tea': {'cost': 40, 'price': 180, 'target_time': 5}
}

items_list = list(menu_items.keys())

# Generate dates and times (heavy focus on the 7 PM - 11 PM dinner rush)
start_date = datetime(2026, 1, 1)
dates = [start_date + timedelta(days=np.random.randint(0, 180)) for _ in range(num_orders)]

for i in range(num_orders):
    # 60% of sales happen during peak dinner hours
    if np.random.random() > 0.4:
        hour = np.random.randint(19, 23)
    else:
        hour = np.random.randint(11, 23)
    dates[i] = dates[i].replace(hour=hour, minute=np.random.randint(0, 60))

# Build the dataset
data = {
    'Ticket_ID': ['TKT-' + str(i).zfill(6) for i in range(1, num_orders + 1)],
    'Order_Timestamp': dates,
    'Location': np.random.choice(locations, num_orders, p=[0.2, 0.35, 0.25, 0.2]),
    'Item_Sold': np.random.choice(items_list, num_orders),
    'Quantity': np.random.choice([1, 2, 3, 4], num_orders, p=[0.6, 0.25, 0.1, 0.05])
}

df = pd.DataFrame(data)

# Calculate financials (Where is the money made or lost?)
df['Unit_Cost'] = df['Item_Sold'].map(lambda x: menu_items[x]['cost'])
df['Unit_Price'] = df['Item_Sold'].map(lambda x: menu_items[x]['price'])
df['Total_Revenue'] = df['Quantity'] * df['Unit_Price']
df['Total_Cost'] = df['Quantity'] * df['Unit_Cost']
df['Gross_Profit'] = df['Total_Revenue'] - df['Total_Cost']

# Calculate Kitchen Bottlenecks (Does the kitchen crash during peak hours?)
def calculate_prep_time(row):
    target = menu_items[row['Item_Sold']]['target_time']
    hour = row['Order_Timestamp'].hour

    # If it is peak time (7 PM - 10 PM), the kitchen gets backed up
    if 19 <= hour <= 22:
        actual = target + np.random.randint(2, 15)
    else:
        actual = target + np.random.randint(-2, 5)

    return actual

df['Actual_Prep_Time_Min'] = df.apply(calculate_prep_time, axis=1)
df['Target_Prep_Time_Min'] = df['Item_Sold'].map(lambda x: menu_items[x]['target_time'])
df['Time_Variance'] = df['Actual_Prep_Time_Min'] - df['Target_Prep_Time_Min']

# Flag tickets that took way too long
df['Is_Late'] = np.where(df['Time_Variance'] > 5, 'Yes', 'No')

df.to_csv('restaurant_operations_25k.csv', index=False)
print("Success! 25,000 rows generated and saved as restaurant_operations_25k.csv")
```
